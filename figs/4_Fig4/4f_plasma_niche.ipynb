```python
FILE_NAME = "adata_combined_new.h5ad.final.filtered"
PATH2=f'/nfs/team298/ls34/adult_skin/final_adatas/{FILE_NAME}'
adata_5k=sc.read_h5ad(PATH2)
adata_5k=adata_5k[adata_5k.obs["tech"]=="xenium"]


def remove_markers(LIST):
    try: 
        LIST = {key: [gene for gene in genes if gene in adata_5k.var_names] 
                           for key, genes in LIST.items()}
    except: 
        LIST =[gene for gene in LIST if gene in adata_5k.var_names]
    return LIST

 
 
```

```python
with open('/nfs/team298/ls34/niche_colors.pkl', 'rb') as f:
    colors_new2 = pickle.load(f)
 
```

```python
sc.settings.set_figure_params(dpi_save=600, facecolor="white", frameon=False, figsize=(7,7))

NICHE =  'Plasma_cell_rich'
MIN_NUMBER_CELLS=10

CORE_PAPER =["BK23_Week 12",
                                       "BK27_Week 12",
                                      'BK27_Lesional Baseline',
                                         'BK24_Lesional Baseline']
for i, tissue_section_id in enumerate(CORE_PAPER):
    print(tissue_section_id)
   # print(i, "/", len(adata_5k.obs["info_id6"].unique()))
    adata_i =  adata_5k[adata_5k.obs["info_id6"]==tissue_section_id]
    niche_names_found = adata_i.obs["niche19"].unique()
    #DISEASE=adata_i.obs["disease_overall"].unique()[0]
    if NICHE in niche_names_found:
        adata_ii =  adata_i[adata_i.obs["niche19"]==NICHE]
        print(tissue_section_id, " - ",  adata_ii.shape[0] , " cells")
        if adata_ii.shape[0] > MIN_NUMBER_CELLS:     
            i=i+1
            print(tissue_section_id, " has ", adata_ii.shape[0])
            adata_i.obs["test"] = [x if x==NICHE else "other" for x in adata_i.obs['niche19']]
            print(adata_i.obs["test"].value_counts())
            
            PC_NICHE_COLOR= colors_new2["Plasma_cell_rich"]
        #"#FFB300" #"#FFD300"
            NUM = ord(NICHE[0])
            if NUM <79:
                custom_palette =ListedColormap([  "#f8f8f8",  PC_NICHE_COLOR]) #NEON B:  "#00e5ff",
            else:
                custom_palette = ListedColormap([   PC_NICHE_COLOR,"#f8f8f8", ])
            adata_i.obs['niche19'] =adata_i.obs['niche19'] .astype('category')
            adata_i.uns['niche19_colors'] = [colors_new2.get(cat, '#D3D3D3')  # Default to light grey if not found
                                     for cat in adata_i.obs['niche19'].cat.categories]

            sq.pl.spatial_scatter(
                    adata_i,
                    library_id="spatial",
                    shape=None,
                    color="test",
                    size=10,
                    vmax=1,
                    title=tissue_section_id,# + "_" + DISEASE,
                    palette=custom_palette,
                    linewidth=0.01,
                    edgecolor="black",
                legend_loc=None,
                #save=f"5a_{tissue_section_id}.pdf"
                save=f'4i_highlighted_{tissue_section_id}.pdf'

                    #ax=ax,
                    #legend_loc="on data"  # Disable the legend for each subplot
                )
            sq.pl.spatial_scatter(
                    adata_i,
                    library_id="spatial",
                    shape=None,
                    color="niche19",
                    size=6,
                    vmax=1,
                    title=tissue_section_id,
                    linewidth=0.1,
                    edgecolor="black",
              legend_loc=None,
                save=f'4i_byniche_{tissue_section_id}.pdf'

                    #ax=ax,
                    #legend_loc="on data"  # Disable the legend for each subplot
                )
#             if i >30:
#                 print("10+ samples found already ... Terminate")
#                 break


#             adata_i.obs["test"] = [x if x == NICHE else "Other" for x in adata_i.obs["niche_name"]]


#             NUM = ord(NICHE[0])
#             if NUM <79:
#                 custom_palette =ListedColormap([  "red",  "lightgrey"])
#             else:
#                 custom_palette = ListedColormap([  "lightgrey", "red"])
#             sq.pl.spatial_scatter(
#                     adata_i,
#                     library_id="spatial",
#                     shape=None,
#                     color="test",
#                     size=20,
#                     vmax=1,
#                     title=tissue_section_id,
#                     palette=custom_palette,
#                     linewidth=0.1,
#                     edgecolor="black",
#                 legend_loc=None

#                     #ax=ax,
#                     #legend_loc="on data"  # Disable the legend for each subplot
#                 )
# #             if i >30:
# #                 print("10+ samples found already ... Terminate")
# #                 break



sc.settings.set_figure_params(dpi=100, facecolor="white", frameon=False, figsize=(18,18))
 
```
